# 1. Input

- Chuỗi N event đúng thứ tự: E1, E2, ..., EN
- Ví dụ:
    - E1: Cắt nấm
    - E2: Cắt củ năng
    - E3: Cắt đậu hũ
    - E4: Bật lửa

# 2. Query encoding

Mỗi event:
- event text => PE-Core text encoder => L2-normalized vector

Output:
- N vectors (u1, u2, ..., uN)

# 3. Similarity

Mỗi event so với visual embedding của keyframe:

`S[i,t] = similarity(event_i, keyframe_t)`

# 4. DANTE

Chạy riêng cho từng video, với keyframe được sắp theo: `metadata.timestamp`. Không được nhảy giữa hai video

DANTE trả:
- video_id
- sequence_score
- sequence keyframes

Logical output: 

```python
{
  "video_id": "V001",
  "sequence_score": 3.42,
  "sequence": [
    {
      "event_id": "E1",
      "frame_id": "frame_1"
    },
    {
      "event_id": "E2",
      "frame_id": "frame_2"
    }
  ]
}
```

Baseline TRAKE hiện chỉ dùng: DANTE + visual_features, không dùng OCR, ASR, Summary, Stable Diffusion, QUEST
