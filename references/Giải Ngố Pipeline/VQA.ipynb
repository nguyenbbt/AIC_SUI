# 1. Input

- mode = VQA
- question
- answer_type

Ví dụ:
- question = "Có bao nhiêu người trên sân khấu?"
- answer_type = "number"

# 2. Query Builder

Sinh:
- vqa_retrieval_rewrite
- paraphrase_1
- paraphrase_2

# 3. Retrieval

Chạy gần giống Textual KIS:
- visual semantic
- OCR lexical
- ASR lexical
- ASR semantic
- summary lexical
- summary semantic
- optional SD/QUEST
- optional object constraints từ UI

# 4. Evidence collection

Từ candidate tốt nhất:
- keyframe images
- near frames
- OCR text
- ASR transcript
- video summary
- timestamp

# 5. Answer generation

question + retrieved evidence => VLM => answer text

- Output: answer (text)
